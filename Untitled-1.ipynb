```python
from pathlib import Path
from zipfile import ZipFile
import numpy as np
import nibabel as nib

# 1. Unzip once (skip this cell after extraction)
zip_path = Path(r"C:\path\to\ASNR-MICCAI-BraTS2023-GLI-Challenge-TrainingData.zip")
data_root = Path(r"C:\path\to\BraTS2023_GLI")

data_root.mkdir(parents=True, exist_ok=True)

if not any(data_root.iterdir()):
    with ZipFile(zip_path, "r") as z:
        z.extractall(data_root)

# 2. Locate patient folders
case_dirs = sorted(p for p in data_root.rglob("BraTS-GLI-*") if p.is_dir())
print(f"{len(case_dirs)} cases found")
print(case_dirs[0])
```

```python
case = case_dirs[0]

files = {
    "T1":    next(case.glob("*-t1n.nii.gz")),
    "T1ce":  next(case.glob("*-t1c.nii.gz")),
    "T2":    next(case.glob("*-t2w.nii.gz")),
    "FLAIR": next(case.glob("*-t2f.nii.gz")),
    "mask":  next(case.glob("*-seg.nii.gz")),
}

# Image shape: (4, depth, height, width)
image = np.stack(
    [nib.load(files[key]).get_fdata(dtype=np.float32)
     for key in ["T1", "T1ce", "T2", "FLAIR"]],
    axis=0,
)

# Label shape: (depth, height, width)
mask = nib.load(files["mask"]).get_fdata(dtype=np.float32).astype(np.uint8)

print(image.shape)
print(mask.shape)
print("Labels:", np.unique(mask))   # Expected: 0, 1, 2, 3
```
